```python
dfs = []
for model, batch_size, in list(product(models, batch_sizes)):
    df_orig = pd.read_csv(f"data/{model}_{batch_size}.csv")
    df_orig = df_orig[df_orig["AP@IoU[0.5:0.95]"] > selection_thresholds[model]]
    print(len(df_orig))

    df = df_orig.sort_values(by=["AP@IoU[0.5:0.95]"], ascending=False)

    dfs.append((model, batch_size, df))
    df.to_csv(f"data/{model}_{batch_size}_sorted_AP.csv", index=False)
```
